```python
def count_within_limits(x_coords, y_coords, xmin, xmax, ymin, ymax):
    count = 0
    for x, y in zip(x_coords, y_coords):
        if xmin <= x <= xmax and ymin <= y <= ymax:
            count += 1
    return count
```

```python
from rasterio.transform import from_origin

def create_geotiff(x_coords, y_coords, pixel_values, output_file, utm_zone=10):
    # Define the UTM zone and hemisphere
    utm_crs = f"EPSG:326{utm_zone:02d}"

    # Calculate the pixel size (assuming equal spacing in x and y directions)
    pixel_size_x = x_coords[1] - x_coords[0]
    pixel_size_y = y_coords[1] - y_coords[0]

    # Calculate the upper-left corner coordinates
    x_min = min(x_coords)
    y_max = max(y_coords)

    # Define the transformation
    transform = from_origin(x_min, y_max, pixel_size_x, pixel_size_y)

    # Create the raster
    with rasterio.open(
        output_file,
        'w',
        driver='GTiff',
        height=len(pixel_values),
        width=len(pixel_values),
        count=1,
        dtype=pixel_values.dtype,
        crs=utm_crs,
        transform=transform
    ) as dst:
        # Write the pixel values to the raster
        dst.write(pixel_values, 1)
```
