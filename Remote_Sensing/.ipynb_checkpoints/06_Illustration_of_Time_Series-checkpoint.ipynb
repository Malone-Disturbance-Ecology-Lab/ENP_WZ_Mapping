```python
import os
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import imageio
import imageio.v3 as iio  # For creating GIFs
```

```python
input_dir = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/"
output_gif = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/Animation_2003_2024_thresholding.gif"

# Step 2: Define color maps
default_colors = ['white', 'red', 'gray', 'green']
default_cmap = ListedColormap(default_colors)

custom_cmaps = {
    "raster_2013": ListedColormap(['white', 'red', 'gray', 'green']),  # 3 → red, 2 → gray, 1 → green
    "raster_2018": ListedColormap(['white', 'red', 'gray', 'green']),  # 3 → gray, 2 → green, 1 → red
    "raster_2023": ListedColormap(['white', 'red', 'gray', 'green']),  # 3 → gray, 2 → red, 1 → green
    ##"raster_2024": ListedColormap(['white', 'red', 'green', 'gray']),
    "raster_2003": ListedColormap(['white', 'red', 'gray', 'green']),
}

bounds = [-0.5, 0.5, 1.5, 2.5, 3.5]
norm = plt.Normalize(vmin=-0.5, vmax=3.5)

# Step 3: Get all GeoTIFF files in the directory
geotiff_files = sorted([os.path.join(input_dir, f) for f in os.listdir(input_dir) if f.endswith('.tif')])

# Step 4: Create a list to store frames for animation
frames = []

for file in geotiff_files:
    with rasterio.open(file) as src:
        data = src.read(1)  # Read the first band
        data[data > 3] = 0  # Mask invalid values, if any

        file_basename = os.path.basename(file).split('.')[0]
        cmap = custom_cmaps.get(file_basename, default_cmap)

        # Plot the data with minimal white space
        fig, ax = plt.subplots(figsize=(6, 6), tight_layout=True)
        ax.imshow(data, cmap=cmap, norm=norm)
        ax.axis('off')  # Remove axes for a clean plot
        plt.subplots_adjust(left=0, right=1, top=1, bottom=0)  # Remove margins

        # Add file name as title
        plt.title(file_basename, fontsize=12, color='black')

        # Save the figure as a numpy array
        fig.canvas.draw()
        frame = np.array(fig.canvas.renderer.buffer_rgba())
        frames.append(frame)
        plt.close(fig)

# Step 5: Save the frames as an animated GIF
iio.imwrite(output_gif, frames, duration=0.5)

print(f"Animation saved to {output_gif}")
```
