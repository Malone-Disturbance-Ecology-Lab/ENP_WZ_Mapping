```python
# 1️⃣ Load CSV
csv_path = "/Users/bz294/Documents/WhiteZone/Validation/MODIS/Samples/MODIS_MIDFEB_LAI_BY_POINT_NEW.csv" 
df = pd.read_csv(csv_path)

# 2️⃣ Remove invalid values
df = df[df['median'] != -9999]
# 2️⃣ Filter only relevant classes
classes = ['TZ', 'WZ', 'VZ']
df = df[df['descriptio'].isin(classes)]

# 3️⃣ Ensure 'year' is numeric
df['year'] = pd.to_numeric(df['year'], errors='coerce')

# 4️⃣ Plot settings
sns.set(style="whitegrid")
palette = {'TZ':'#1f77b4', 'WZ':'#ff7f0e', 'VZ':'#2ca02c'}

```

```python
def remove_outliers(group):
    Q1 = group['median'].quantile(0.25)
    Q3 = group['median'].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return group[(group['median'] >= lower) & (group['median'] <= upper)]
```

```python
df = df.groupby('descriptio', group_keys=False).apply(remove_outliers)

# 6️⃣ Plot settings
sns.set(style="whitegrid")
palette = {'TZ':'#1f77b4', 'WZ':'#ff7f0e', 'VZ':'#2ca02c'}

# ---------------------------------------------
# 7️⃣ Distribution per year
years = sorted(df['year'].unique())

for year in years:
    plt.figure(figsize=(8,5))
    df_year = df[df['year']==year]
    
    sns.violinplot(x='descriptio', y='median', data=df_year,
                   inner=None, palette=palette, cut=0)
    sns.boxplot(x='descriptio', y='median', data=df_year,
                whis=[0,100], width=0.1, palette=palette, fliersize=0)
    
    # Add median markers
    medians = df_year.groupby('descriptio')['median'].median()
    for i, cls in enumerate(classes):
        plt.scatter(i, medians[cls], color='black', s=50, zorder=10, marker='D')
    
    plt.title(f"Distribution of Median LAI Values ({year})")
    plt.ylabel("Median LAI")
    plt.xlabel("Class")
    plt.tight_layout()
    plt.show()

# ---------------------------------------------
# 8️⃣ Distribution across all years
plt.figure(figsize=(8,5))
sns.violinplot(x='descriptio', y='median', data=df,
               inner=None, palette=palette, cut=0)
sns.boxplot(x='descriptio', y='median', data=df,
            whis=[0,100], width=0.1, palette=palette, fliersize=0)

# Median markers
medians = df.groupby('descriptio')['median'].median()
for i, cls in enumerate(classes):
    plt.scatter(i, medians[cls], color='black', s=50, zorder=10, marker='D')

plt.title("Distribution of Median LAI Values (All Years, Outliers Removed)")
plt.ylabel("Median LAI")
plt.xlabel("Class")
plt.tight_layout()
plt.show()
```

```python
##### Extract the pixels with a valid lai value #####
# Filter rows where median is valid
df_valid = df[df["median"] != -9999]

# (Optional) keep only specific columns
df_valid = df_valid[["year", "lat", "lon", "median", "descriptio"]]

# Write to new CSV
df_valid.to_csv("/Users/bz294/Documents/WhiteZone/Validation/filtered_by_year_lat_lon.csv", index=False)
```

```python
### Read in the rasters and detect homogeneity #####
raster_dir = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/CDthreshold06/CD_closest_GE"
# -----------------------------
# Detect raster files by year
# -----------------------------
raster_by_year = {}
pattern = re.compile(r"raster_(\d{4})")

for fname in os.listdir(raster_dir):
    match = pattern.search(fname)
    if match:
        year = int(match.group(1))
        raster_by_year[year] = os.path.join(raster_dir, fname)

print(f"Detected rasters for years: {sorted(raster_by_year.keys())}")
```

```python
### Threshold and function to select pixels #####

BUFFER_M = 500
THRESH = 0.6
THRESH_WZ = 0.54  # threshold for WZ

CLASS_MAP = {"TZ": 3, "VZ": 1, "WZ": 2}

transformer = Transformer.from_crs(
    "EPSG:4326",
    "EPSG:32617",
    always_xy=True
)

def is_good_pixel(row):
    year = int(row["year"])
    if year not in raster_by_year:
        return False

    target_class = CLASS_MAP.get(row["descriptio"])
    if target_class is None:
        return False

    # Select threshold based on class
    threshold = THRESH_WZ if row["descriptio"] == "WZ" else THRESH

    raster_path = raster_by_year[year]

    # Reproject point to UTM 17N
    x, y = transformer.transform(row["lon"], row["lat"])
    point = Point(x, y)

    with rasterio.open(raster_path) as src:
        row_px, col_px = src.index(x, y)

        pixel_size = src.res[0]
        radius_px = int(BUFFER_M / pixel_size)

        window = Window(
            col_px - radius_px,
            row_px - radius_px,
            radius_px * 2,
            radius_px * 2
        )

        data = src.read(1, window=window, masked=True)

        if data.mask.all():
            return False

        # Build circular mask
        transform = src.window_transform(window)
        mask = geometry_mask(
            [point.buffer(BUFFER_M)],
            transform=transform,
            invert=True,
            out_shape=data.shape
        )

        values = data[mask]
        values = values[values > 0]  # optional: ignore nodata

        if len(values) == 0:
            return False

        frac = np.sum(values == target_class) / len(values)

        return frac >= threshold
```

```python
df_valid["good_pixel"] = df_valid.apply(is_good_pixel, axis=1)

df_good = df_valid[df_valid["good_pixel"]].copy()

print(f"Homogeneous pixels: {len(df_good)}")
df_good.to_csv("/Users/bz294/Documents/WhiteZone/Validation/MODIS/Samples/homogeneous_pixels_utm17_500m.csv", index=False)
```

```python
### Plot only for the good points satisfying the requirements ######
%matplotlib agg
# Set Times New Roman font and larger size
plt.rcParams.update({'font.family': 'Times New Roman', 'font.size': 20})
fs = 20
# New class order and names
classes = ["DZ", "TZ", "VZ"]
original_classes = ["WZ", "TZ", "VZ"]  # mapping from original

# Prepare the data in the new order
data_to_plot = [df_good[df_good["descriptio"] == orig]["median"] for orig in original_classes]

## FDR ##
# Calculate mean FDR among the three classes
class_means = [d.mean() for d in data_to_plot]
class_vars = [d.var(ddof=1) for d in data_to_plot]

# Generalized Fisher ratio (mean FDR)
fdr = np.var(class_means, ddof=1) / np.mean(class_vars)


plt.figure(figsize=(8, 6))

# Create box plot with colored boxes
box = plt.boxplot(data_to_plot, labels=classes, patch_artist=True, notch=True, showfliers=False)

# Optional: set custom colors for each class
colors = ["#b08b5b", "#a1d996", "#00441b"]  # match DZ, TZ, VZ
for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)

plt.xlabel("Class", fontsize = fs)
plt.ylabel("LAI", fontsize = fs)
plt.title("Box Plot of Median Values by Class", fontsize = fs)
plt.xticks(fontsize=fs)
plt.yticks(fontsize=fs)
plt.tight_layout()

# Save as TIFF with white background
output_path = "/Users/bz294/Documents/WhiteZone/Validation/MODIS/Samples/median_boxplot_point_samples.tif"
plt.savefig(output_path, format='tiff', dpi=300, facecolor='white')
plt.show()
```

```python
# -----------------------------
# Paths
# -----------------------------
csv_file = "/Users/bz294/Documents/WhiteZone/Validation/MODIS/Spatial/MODIS_SPATIAL_.csv"
raster_dir = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/CDthreshold06/CD_closest_GE"


# -----------------------------
# Read CSV
# -----------------------------
df = pd.read_csv(csv_file)

# -----------------------------
# Function to extract dominant class in 500x500 m window
# -----------------------------
def evaluate_pixel(lat_deg, lon_deg, raster_path):
    """
    Extract 500x500 m window around lat/lon (in degrees) after transforming to raster CRS,
    compute class proportions, and return if it's a good pixel and the dominant class.
    """
    with rasterio.open(raster_path) as src:
        # Setup transformer from WGS84 to raster CRS
        transformer = Transformer.from_crs("epsg:4326", src.crs, always_xy=True)
        lon, lat = transformer.transform(lon_deg, lat_deg)
        
        # Define 500x500 m bounding box in raster CRS
        half_size = 250  # meters
        minx = lon - half_size
        maxx = lon + half_size
        miny = lat - half_size
        maxy = lat + half_size

        # Convert bounding box to raster window
        window = from_bounds(minx, miny, maxx, maxy, src.transform)
        
        # Read data in window
        try:
            data = src.read(1, window=window)
        except:
            return False, None

        # Flatten and remove nodata
        data_flat = data.flatten()
        if src.nodata is not None:
            data_flat = data_flat[data_flat != src.nodata]
        if len(data_flat) == 0:
            return False, None

        # Compute class proportions
        unique, counts = np.unique(data_flat, return_counts=True)
        total = counts.sum()
        proportions = {k: v / total for k, v in zip(unique, counts)}

        # Determine if any class > 0.8
        for cls, prop in proportions.items():
            if prop >= 0.8:
                return True, int(cls)

        return False, None

# -----------------------------
# Evaluate all points
# -----------------------------
results = []

for idx, row in df.iterrows():
    lat = row['lat']
    lon = row['lon']
    year = int(row['year'])
    lai = row['Lai_500m']

    raster_path = raster_by_year.get(year)
    if raster_path is None:
        continue  # skip if raster not available

    good, dom_class = evaluate_pixel(lat, lon, raster_path)
    if good:
        results.append({
            'year': year,
            'lat': lat,
            'lon': lon,
            'dominant_class': dom_class,
            'Lai_500m': lai
        })

# -----------------------------
# Save results
# -----------------------------
df_good = pd.DataFrame(results)
df_good.to_csv("/Users/bz294/Documents/WhiteZone/Validation/MODIS/Spatial/MODIS_SPATIAL_good_pixels.csv", index=False)
print(f"Saved {len(df_good)} good pixels")
```

```python
# -----------------------------
# Set Times New Roman font and larger size
# -----------------------------
plt.rcParams.update({'font.family': 'Times New Roman', 'font.size': 20})
fs = 20

# -----------------------------
# Map dominant_class numbers to names
# 1 -> VZ, 2 -> DZ, 3 -> TZ
# -----------------------------
class_mapping = {1: "VZ", 2: "DZ", 3: "TZ"}

# Desired order on X-axis: DZ, TZ, VZ
x_order = [2, 3, 1]
colors = ["#b08b5b", "#a1d996", "#00441b"]  # match DZ, TZ, VZ

# -----------------------------
# Prepare data in the correct order
# -----------------------------
data_to_plot = [df_good[df_good["dominant_class"] == c]["Lai_500m"] for c in x_order]
labels = [class_mapping[c] for c in x_order]

# -----------------------------
# Create boxplot without extreme values (fliers)
# -----------------------------
plt.figure(figsize=(8, 6))
box = plt.boxplot(
    data_to_plot,
    labels=labels,
    patch_artist=True,
    notch=True,
    showfliers=False  # hide the outliers
)

# Set custom colors for each box
for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)

# -----------------------------
# Labels and title
# -----------------------------
plt.xlabel("Class", fontsize=fs)
plt.ylabel("LAI", fontsize=fs)
plt.title("Box Plot of LAI by Dominant Class (Good Pixels)", fontsize=fs)
plt.xticks(fontsize=fs)
plt.yticks(fontsize=fs)
plt.tight_layout()

# -----------------------------
# Save as TIFF with white background
# -----------------------------
output_path = "/Users/bz294/Documents/WhiteZone/Validation/MODIS/Spatial/median_boxplot.tif"
plt.savefig(output_path, format='tiff', dpi=300, facecolor='white')
plt.show()
```
