```python
###################### Generating conceptual framework plots #######################
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 14  # Set a larger base font size
```

```python
%matplotlib agg
# Generate synthetic data with positive trends and higher noise
np.random.seed(1)
n_points = 100

# Independent variables
salinity = np.sort(np.random.uniform(0, 35, n_points))
water_level = np.sort(np.random.uniform(0, 2, n_points))  # in meters

# Dependent variables (Whitezone area in hectares), with noise
wz_area_salinity = 10 * salinity + np.random.normal(0, 30, n_points)  # noisier
wz_area_water = 200 * water_level + np.random.normal(0, 50, n_points)

# Clip to non-negative area values
wz_area_salinity = np.clip(wz_area_salinity, 0, None)
wz_area_water = np.clip(wz_area_water, 0, None)

# Create plots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Whitezone area vs salinity
axes[0].scatter(salinity, wz_area_salinity, color='blue', alpha=0.7)
axes[0].set_xlabel('Salinity', fontsize=16)
axes[0].set_ylabel('WZ Area', fontsize=16)
axes[0].set_title('WZ Area vs Salinity\n(Positive Trend)', fontsize=18)
axes[0].tick_params(labelsize=14)
axes[0].set_xticklabels([])
axes[0].set_yticklabels([])



# Plot 2: Whitezone area vs water level
axes[1].scatter(water_level, wz_area_water, color='green', alpha=0.7)
axes[1].set_xlabel('Water Level', fontsize=16)
axes[1].set_ylabel('WZ Area', fontsize=16)
axes[1].set_title('WZ Area vs Water Level\n(Positive Trend)', fontsize=18)
axes[1].tick_params(labelsize=14)
axes[1].set_xticklabels([])
axes[1].set_yticklabels([])

# Save and display
plt.tight_layout()
plt.savefig("/Users/bz294/Documents/WhiteZone/Figure/whitezone_scatter_plots.tif", format='tiff', dpi=300)
plt.show()
```

```python
%matplotlib agg

# Set font
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 14

# Generate synthetic data
np.random.seed(2)
n_points = 100

# Independent variables
salinity = np.sort(np.random.uniform(0, 35, n_points))
water_level = np.sort(np.random.uniform(0, 2, n_points))  # in meters

# WZ Area increases with water level (positive) and decreases with salinity (negative)
wz_area_salinity = 8 * salinity + np.random.normal(0, 30, n_points)  # decreasing trend
wz_area_water = 500 - 150 * water_level + np.random.normal(0, 40, n_points)      # increasing trend

# Clip to non-negative area values
wz_area_salinity = np.clip(wz_area_salinity, 0, None)
wz_area_water = np.clip(wz_area_water, 0, None)

# Create plots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: WZ area vs salinity (negative trend)
axes[0].scatter(salinity, wz_area_salinity, color='blue', alpha=0.7)
axes[0].set_xlabel('Salinity (PSU)', fontsize=16)
axes[0].set_ylabel('WZ Area', fontsize=16)
axes[0].set_title('WZ Area vs Salinity\n(Positive Trend)', fontsize=18)
axes[0].tick_params(labelsize=14)
axes[0].set_xticklabels([])
axes[0].set_yticklabels([])
# Plot 2: WZ area vs water level (positive trend)
axes[1].scatter(water_level, wz_area_water, color='green', alpha=0.7)
axes[1].set_xlabel('Water Level (m)', fontsize=16)
axes[1].set_ylabel('WZ Area', fontsize=16)
axes[1].set_title('WZ Area vs Water Level\n(Negative Trend)', fontsize=18)
axes[1].tick_params(labelsize=14)
axes[1].set_xticklabels([])
axes[1].set_yticklabels([])
plt.xticks([]) 
plt.yticks([])

# Save the figure
plt.tight_layout()
plt.savefig("/Users/bz294/Documents/WhiteZone/Figure/whitezone_opposing_trends.tif", format='tiff', dpi=300)
plt.show()
```
