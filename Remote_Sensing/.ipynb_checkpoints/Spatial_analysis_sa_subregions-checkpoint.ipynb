```python
## De-mean ######

# Filter years 2013–2024
test_df = salinity_df[(salinity_df["year"] >= 2014) & (salinity_df["year"] <= 2024)].copy()

# Subtract station mean instead of 2013 baseline
demeaned_list = []
for station, subdf in test_df.groupby("site"):
    subdf = subdf.copy()
    mean_val = subdf["annual_avg"].mean()
    subdf["demeaned"] = subdf["annual_avg"] - mean_val
    demeaned_list.append(subdf)

demeaned_df = pd.concat(demeaned_list)

# Assign region
def assign_region(site):
    if site in west:
        return "west"
    elif site in central:
        return "central"
    else:
        return "east"

demeaned_df["region"] = demeaned_df["site"].apply(assign_region)

# Calculate slope, p-value, r2 for each region
trends = {}
for region, subdf in demeaned_df.groupby("region"):
    yearly_mean = subdf.groupby("year")["demeaned"].mean().reset_index()
    slope, intercept, r_value, p_value, std_err = linregress(yearly_mean["year"], yearly_mean["demeaned"])
    trends[region] = {"slope": slope, "p_value": p_value, "r2": r_value**2}

trend_df = pd.DataFrame(trends).T
print("Demeaned Salinity Trends (2013–2024):")
print(trend_df)
```

```python
groups = {'West': west, 'Central': central, 'East': east}
def group_corr_stats(df, group_sites):
    # Keep only sites in this group
    df_group = df[df['site'].isin(group_sites)].copy()
    
    # Aggregate to remove duplicates (average if multiple rows for same year & site)
    df_group = df_group.groupby(['year', 'site'], as_index=False)['annual_avg'].mean()
    
    # Pivot to have years as index, sites as columns
    df_pivot = df_group.pivot(index='year', columns='site', values='annual_avg')
    
    # Keep only years where all sites have data → common time span
    df_common = df_pivot.dropna(how='any')
    
    common_years = df_common.index.tolist()
    
    if len(df_common.columns) < 2 or df_common.empty:
        return None, None, common_years
    
    # Correlation matrix
    corr_matrix = df_common.corr()
    
    # Extract correlations excluding diagonal (self-correlation = 1)
    corr_values = corr_matrix.values[np.triu_indices_from(corr_matrix.values, k=1)]
    
    corr_mean = corr_values.mean()
    corr_std = corr_values.std()
    
    return corr_mean, corr_std, common_years


```

```python
groups = {'West': west, 'Central': central, 'East': east}

for region, sites in groups.items():
    corr_mean, corr_std, common_years = group_corr_stats(salinity_df, sites)
    print(f"\n=== {region} Region ===")
    if common_years:
        print(f"Common years: {common_years[0]} – {common_years[-1]} ({len(common_years)} years)")
        if corr_mean is not None:
            print(f"Correlation among sites: mean = {corr_mean:.3f}, std = {corr_std:.3f}")
        else:
            print("Not enough data to calculate correlations.")
    else:
        print("No common years with complete data!")
```

```python
# Combine all sites
all_sites = west + central + east

# Calculate correlation stats across all sites
corr_mean_all, corr_std_all, common_years_all = group_corr_stats(salinity_df, all_sites)

print(f"\n=== All Sites Combined ===")
if common_years_all:
    print(f"Common years: {common_years_all[0]} – {common_years_all[-1]} ({len(common_years_all)} years)")
    if corr_mean_all is not None:
        print(f"Correlation among sites: mean = {corr_mean_all:.3f}, std = {corr_std_all:.3f}")
    else:
        print("Not enough data to calculate correlations.")
else:
    print("No common years with complete data!")
```

```python
#### Process the lat and lon file for salinity ####
import pandas as pd
import numpy as np

csv_path = "/Volumes/Malonelab/Research/Whitezone/Hydro_Salinity/Salinity/WZ_salinity_sites_latlon.csv"

df = pd.read_csv(csv_path)

def ddmmss_to_degree(value):
    if pd.isna(value):
        return np.nan

    value = float(value)

    degrees = int(value // 10000)
    minutes = int((value % 10000) // 100)
    seconds = value % 100

    return degrees + minutes / 60.0 + seconds / 3600.0

# =========================
# Apply conversion
# =========================
df["LAT_deg"] = df["LAT (ddmmss.sss)"].apply(ddmmss_to_degree)
df["LON_deg"] = -df["LON (ddmmss.sss)"].apply(ddmmss_to_degree)

output_path = "/Volumes/Malonelab/Research/Whitezone/Hydro_Salinity/Salinity/WZ_salinity_sites_latlon_deg_.csv"
df.to_csv(output_path, index=False)

print("Conversion completed.")

```

```python
def monthly_average_all_clean(df):
    """
    Calculate monthly average for all stations, removing rows with missing data.
    
    Assumes:
    - 'date' column is datetime or convertible
    - 'value' column may be string, will be converted to numeric
    - 'station' column exists
    """
    if df.empty:
        print("⚠️ The dataframe is empty")
        return pd.DataFrame()
    
    # ensure date column is datetime
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date"])
    
    # convert 'value' to numeric, coerce errors to NaN
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    
    # drop rows where value is NaN
    df = df.dropna(subset=["value"])
    
    # set date as index
    df.set_index("date", inplace=True)
    
    # resample monthly per station and calculate mean
    monthly_df = (
        df.groupby("station")["value"]
        .resample("M")
        .mean()
        .dropna()  # remove months with no data
        .reset_index()
    )
    
    return monthly_df

# Apply to west, central, east
west_monthly = monthly_average_all_clean(west_df)
central_monthly = monthly_average_all_clean(central_df)
east_monthly = monthly_average_all_clean(east_df)

print("✅ Monthly averages for all stations calculated and cleaned")
print(west_monthly.head())
```

```python
########### Check elevated salinity events ###########

%matplotlib agg

def monthly_exceedances(df, threshold=40, start_year=1991, end_year=2024):
    """
    For each year, calculate the average number of months across stations
    where the monthly average salinity exceeds `threshold`.
    """
    df = df.copy()
    
    # Ensure numeric and datetime
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    df = df.dropna(subset=["value"])
    
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date"])
    
    # Keep only years in range
    df["year"] = df["date"].dt.year
    df = df[(df["year"] >= start_year) & (df["year"] <= end_year)]
    
    # Flag months above threshold
    df["above_thresh"] = df["value"] > threshold
    
    # Count months > threshold per year per station
    monthly_counts = df.groupby(["station", "year"])["above_thresh"].sum().reset_index()
    
    # Average across stations for each year
    yearly_avg_months = monthly_counts.groupby("year")["above_thresh"].mean().reset_index()
    yearly_avg_months.rename(columns={"above_thresh": "avg_months_above_thresh"}, inplace=True)
    
    return yearly_avg_months

# Compute for each region
west_yearly = monthly_exceedances(west_monthly)
central_yearly = monthly_exceedances(central_monthly)
east_yearly = monthly_exceedances(east_monthly)

# Plotting
plt.figure(figsize=(12,6))

plt.plot(west_yearly["year"], west_yearly["avg_months_above_thresh"], label="West", color="#fee08b", marker="o")
plt.plot(central_yearly["year"], central_yearly["avg_months_above_thresh"], label="Central", color="orange", marker="o")
plt.plot(east_yearly["year"], east_yearly["avg_months_above_thresh"], label="East", color="red", marker="o")

plt.xticks([1991, 2002, 2013, 2024], fontsize=20)  # set xtick font size
plt.yticks(fontsize = 20)
plt.xlabel("Year", fontsize=20)
plt.ylabel("Number of months with salinity > 40 PSU", fontsize=20)
#plt.title("", fontsize=20)
plt.legend(fontsize=20)
plt.grid(True)
plt.tight_layout()
plt.show()
output_dir = "/Users/bz294/Documents/WhiteZone/Figure/Sup"
output_path = os.path.join(output_dir, "salinity_months.png")
plt.savefig(output_path, dpi=300)  # dpi=300 for publication quality
```

```python
def days_over_40_per_year(df):
    """
    Calculate yearly average number of days with daily salinity > 40 PSU across all stations.
    
    df: daily-resolution dataframe with columns ['station', 'date', 'value']
    """
    df = df.copy()
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df = df.dropna(subset=["date", "value"])
    
    # Ensure numeric
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    df = df.dropna(subset=["value"])
    
    # Extract year
    df["year"] = df["date"].dt.year
    
    # Flag days > 40
    df["over_40"] = df["value"] > 40
    
    # Count number of days > 40 per year per station
    days_per_station = (
        df.groupby(["station", "year"])["over_40"]
        .sum()
        .reset_index()
        .rename(columns={"over_40": "days_over_40"})
    )
    
    # Average across stations per year
    avg_days_per_year = (
        days_per_station.groupby("year")["days_over_40"]
        .mean()
        .reset_index()
        .rename(columns={"days_over_40": "avg_days_over_40"})
    )
    
    # Keep only years 1991–2024
    avg_days_per_year = avg_days_per_year[(avg_days_per_year["year"] >= 1991) & (avg_days_per_year["year"] <= 2024)]
    
    return avg_days_per_year

# Apply to each subregion
west_ts = days_over_40_per_year(west_enp)
central_ts = days_over_40_per_year(central_enp)

# Plot
plt.figure(figsize=(12, 6))
plt.plot(west_ts["year"], west_ts["avg_days_over_40"], label="West", marker="o", color="#fee08b", linewidth=2.5, markersize=10   )
plt.plot(central_ts["year"], central_ts["avg_days_over_40"], label="Central", marker="s", color="orange")
#plt.plot(east_ts["year"], east_ts["avg_days_over_40"], label="East", marker="^", color="red")

plt.xlim(1991, 2024)  # Force x-axis to start at 1991
plt.xticks([1991, 2002, 2013, 2024], fontsize=28)
plt.yticks(fontsize=28)
plt.xlabel("Year", fontsize=28)
plt.ylabel("Number of days with salinity > 40 PSU", fontsize=28)
#plt.title("Time Series of Days per Year with Salinity > 40 (1991–2024)", fontsize=20)
#plt.legend(fontsize=16)
plt.grid(True)
plt.tight_layout()

# Save figure
output_dir = "/Users/bz294/Documents/WhiteZone/Figure/Sup"
plt.savefig(f"{output_dir}/days_over_40_timeseries.png", dpi=300)
plt.show()
```

```python
#### Using the original df to extract percentage of exceedance #####

def yearly_percent_over_40(df, start_year=1991, end_year=2024):
    df = df.copy()
    
    # Clean types
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df["value"] = pd.to_numeric(df["value"], errors="coerce")
    df = df.dropna(subset=["date", "value"])
    
    # Year filter
    df["year"] = df["date"].dt.year
    df = df[(df["year"] >= start_year) & (df["year"] <= end_year)]
    
    # Flag exceedance
    df["over_40"] = df["value"] > 40
    
    # Compute percentage per year
    yearly = (
        df.groupby("year")
          .agg(total_rows=("value", "size"),
               over_rows=("over_40", "sum"))
          .reset_index()
    )
    
    yearly["percent_over_40"] = yearly["over_rows"] / yearly["total_rows"] * 100
    
    return yearly

# Filter all three subregions for stations starting with "FLAB"
west_df_filtered = west_df[west_df["station"].str.startswith("FLAB")]
central_df_filtered = central_df[central_df["station"].str.startswith("FLAB")]
east_df_filtered = east_df[east_df["station"].str.startswith("FLAB")]

# Compute yearly percentages
west_pct = yearly_percent_over_40(west_df_filtered)
central_pct = yearly_percent_over_40(central_df_filtered)
east_pct = yearly_percent_over_40(east_df_filtered)
```

```python
plt.figure(figsize=(12,6))

plt.plot(west_pct["year"], west_pct["percent_over_40"],
         label="West", marker="o", color = "#fee08b")

plt.plot(central_pct["year"], central_pct["percent_over_40"],
         label="Central", marker="s", color = "orange")

plt.plot(east_pct["year"], east_pct["percent_over_40"],
         label="East", marker="^", color = "red")

plt.xticks([1991, 2002, 2013, 2024], fontsize=20)
plt.yticks(fontsize=20)

plt.xlabel("Year", fontsize=20)
plt.ylabel("Percent of samples with salinity>40 PSU (%)", fontsize=20)
plt.title("",
          fontsize=20)

plt.legend(fontsize=20)
plt.grid(True)
plt.tight_layout()

# Save figure
output_dir = "/Users/bz294/Documents/WhiteZone/Figure/Sup"
os.makedirs(output_dir, exist_ok=True)
plt.savefig(os.path.join(output_dir, "percent_days_over40_timeseries.png"), dpi=300)

plt.show()
```
