```python
import os
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import imageio
import imageio.v3 as iio  # For creating GIFs
```

```python
input_dir = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/"
output_gif = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/Animation_2003_2024_thresholding.gif"

# Step 2: Define color maps
default_colors = ['white', 'red', 'gray', 'green']
default_cmap = ListedColormap(default_colors)

custom_cmaps = {
    "raster_2013": ListedColormap(['white', 'red', 'gray', 'green']),  # 3 → red, 2 → gray, 1 → green
    "raster_2018": ListedColormap(['white', 'red', 'gray', 'green']),  # 3 → gray, 2 → green, 1 → red
    "raster_2023": ListedColormap(['white', 'red', 'gray', 'green']),  # 3 → gray, 2 → red, 1 → green
    ##"raster_2024": ListedColormap(['white', 'red', 'green', 'gray']),
    "raster_2003": ListedColormap(['white', 'red', 'gray', 'green']),
}

bounds = [-0.5, 0.5, 1.5, 2.5, 3.5]
norm = plt.Normalize(vmin=-0.5, vmax=3.5)

# Step 3: Get all GeoTIFF files in the directory
geotiff_files = sorted([os.path.join(input_dir, f) for f in os.listdir(input_dir) if f.endswith('.tif')])

# Step 4: Create a list to store frames for animation
frames = []

for file in geotiff_files:
    with rasterio.open(file) as src:
        data = src.read(1)  # Read the first band
        data[data > 3] = 0  # Mask invalid values, if any

        file_basename = os.path.basename(file).split('.')[0]
        cmap = custom_cmaps.get(file_basename, default_cmap)

        # Plot the data with minimal white space
        fig, ax = plt.subplots(figsize=(6, 6), tight_layout=True)
        ax.imshow(data, cmap=cmap, norm=norm)
        ax.axis('off')  # Remove axes for a clean plot
        plt.subplots_adjust(left=0, right=1, top=1, bottom=0)  # Remove margins

        # Add file name as title
        plt.title(file_basename, fontsize=12, color='black')

        # Save the figure as a numpy array
        fig.canvas.draw()
        frame = np.array(fig.canvas.renderer.buffer_rgba())
        frames.append(frame)
        plt.close(fig)

# Step 5: Save the frames as an animated GIF
iio.imwrite(output_gif, frames, duration=0.5)

print(f"Animation saved to {output_gif}")
```

```python
#### the plot with pie chart #####################
import os
import numpy as np
import rasterio
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from matplotlib import rcParams
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

# Set global font to Times New Roman
rcParams['font.family'] = 'Times New Roman'
rcParams['font.size'] = 18

# Set up paths
input_dir = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/CDthreshold06/CD_closest_GE"
output_figure = os.path.join(input_dir, "LandCover_3x3_Figure_with_Pies.tif")

# Define colors and categories
class_colors = ['#2b83ba', '#d7191c', '#ffffbf']  # 1 - VZ, 2 - WZ, 3 - TZ
default_colors = ['white'] + class_colors  # include 0 (white)
default_cmap = ListedColormap(default_colors)
norm = plt.Normalize(vmin=-0.5, vmax=3.5)

# Years to include
target_years = [1985, 1994, 1999, 2003, 2009, 2013, 2018, 2020, 2024]
file_map = {}

# Match files to years
for f in os.listdir(input_dir):
    if f.endswith('.tif'):
        base = os.path.splitext(f)[0]
        for year in target_years:
            if str(year) in base:
                file_map[year] = os.path.join(input_dir, f)

# Sort years and create the 3x3 figure
sorted_years = sorted(file_map.keys())
fig, axes = plt.subplots(3, 3, figsize=(13, 13))
axes = axes.flatten()

for i, year in enumerate(sorted_years):
    file_path = file_map[year]
    with rasterio.open(file_path) as src:
        data = src.read(1)
        data[data > 3] = 0  # mask invalid values

        axes[i].imshow(data, cmap=default_cmap, norm=norm)

        # Add custom title
        title_text = f"{year} (non-GE year)" if year >= 2020 else str(year)
        axes[i].set_title(title_text, fontsize=20, fontname='Times New Roman')
        axes[i].axis('off')

        # Compute percentages for classes 1, 2, 3
        total_valid = np.sum(np.isin(data, [1, 2, 3]))
        class_counts = [
            np.sum(data == 1),
            np.sum(data == 2),
            np.sum(data == 3)
        ]
        class_percents = [count / total_valid for count in class_counts] if total_valid > 0 else [0, 0, 0]

        # Create inset pie chart
        inset_ax = inset_axes(axes[i], width="35%", height="35%", loc='lower right', borderpad=1)
        inset_ax.pie(
            class_percents,
            colors=class_colors,
            startangle=90,
            wedgeprops=dict(linewidth=0.5, edgecolor='gray')
        )
        inset_ax.set_aspect('equal')
        inset_ax.axis('off')

plt.tight_layout()
plt.savefig(output_figure, dpi=300, format='tiff')
plt.close()

print(f"3x3 figure with pie charts saved to {output_figure}")
```

```python
import os
import numpy as np
import rasterio
import matplotlib.pyplot as plt
from matplotlib import rcParams
import matplotlib.patches as patches

# Font settings
rcParams['font.family'] = 'Times New Roman'
rcParams['font.size'] = 16

# Years to highlight with red border
highlight_years = {'1985', '1994', '1999', '2003', '2009', '2013', '2018', '2020', '2023'}

# Paths
input_dir = "/Volumes/Malonelab/Research/Whitezone/Result/Emulation/Final/CDthreshold06/CD_closest_GE"
output_figure = os.path.join(input_dir, "TransitionZone_TimeSeries.tif")

# Get and sort GeoTIFFs, excluding the output file
geotiff_files = sorted([
    os.path.join(input_dir, f)
    for f in os.listdir(input_dir)
    if f.endswith('.tif') and f != os.path.basename(output_figure)
])

# Layout: 7 columns × 6 rows
ncols = 7
nrows = 6
total_slots = ncols * nrows

# Create figure
fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(ncols * 2.5, nrows * 2.5))
axes = axes.flatten()

for i in range(total_slots):
    ax = axes[i]

    if i < len(geotiff_files):
        file = geotiff_files[i]
        with rasterio.open(file) as src:
            data = src.read(1)
            mask = (data == 3).astype(np.uint8)  # Class 2 only

        ax.imshow(mask, cmap='gray_r', vmin=0, vmax=1)

        # Clean title
        title_raw = os.path.splitext(os.path.basename(file))[0]
        title_clean = title_raw.replace('raster_', '').replace('_cd', '')

        # Try to extract year as int
        try:
            year = int(''.join(filter(str.isdigit, title_clean)))
        except ValueError:
            year = None

        # Assign background color based on year range
        if year is not None:
            if 1984 <= year <= 2003:
                bg_color = 'lightblue'
            elif 2004 <= year <= 2013:
                bg_color = 'lightgreen'
            else:
                bg_color = 'indianred'
        else:
            bg_color = 'lightgray'  # fallback

        ax.set_title(title_clean, fontsize=18, backgroundcolor=bg_color)

        # Red border for key years
        if any(y in title_clean for y in highlight_years):
            rect = patches.Rectangle(
                (0, 0), 1, 1, transform=ax.transAxes,
                linewidth=4, edgecolor='red', facecolor='none'
            )
            ax.add_patch(rect)
    else:
        ax.axis('off')  # Empty slots

    ax.axis('off')

plt.tight_layout()
plt.savefig(output_figure, dpi=300, format='tiff')
plt.close()

print(f"White zone class time series saved to {output_figure}")
```
