```python
## generating time line marking GE-year and non-GE years

import matplotlib.pyplot as plt
import os
```

```python
%matplotlib agg
fs = 24
# Set font to Times New Roman
plt.rcParams['font.family'] = 'Times New Roman'

# Timeline years
years = list(range(1984, 2025))
blue_dots = [1985, 1994, 1999, 2003, 2009, 2013, 2018, 2023]
red_star = [2005]

# Create timeline plot
plt.figure(figsize=(12, 2))
plt.hlines(1, min(years), max(years), colors='gray', linestyles='solid')

# Plot blue dots
plt.plot(blue_dots, [1]*len(blue_dots), 'bo', label='Key Years')

# Plot red star
plt.plot(red_star, [1]*len(red_star), 'r*', markersize=fs, label='Special Year')

# Add year labels for blue dots and red star
for year in blue_dots:
    plt.text(year, 1.02, str(year), ha='center', fontsize=fs, rotation=45)
plt.text(red_star[0], 1.02, str(red_star[0]), ha='center', fontsize=fs, rotation=45, color='red')

# Only label 1984 and 2024 on x-axis
plt.xticks([1984, 2024], fontsize=fs)
plt.xlim(1983, 2025)

# Remove y-axis
plt.yticks([])

# Set plot limits and labels
plt.ylim(0.95, 1.1)
plt.xlabel("Year", fontsize=18)

# Remove the outer border
for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()

# Save to specific directory
output_path = "/Users/bz294/Documents/WhiteZone/Figure/Method_timeline_plot.tif"  # Change this to your path
plt.savefig(output_path, dpi=300, bbox_inches='tight', format='tiff')
```

```python
%matplotlib agg
fs = 22
# Set font to Times New Roman
plt.rcParams['font.family'] = 'Times New Roman'

# Timeline and key years
years = list(range(1984, 2025))
blue_dots = [1985, 1994, 1999, 2003, 2009, 2013, 2018, 2023]

# Define Landsat periods (non-overlapping, blue dots included)
ls5_period = (1984, 2003.2)     # Covers 2003
ls7_period = (2003.2, 2013.2)   # Covers 2013
ls8_period = (2013.2, 2024)

# Create the plot
plt.figure(figsize=(12, 2))
plt.hlines(1, min(years), max(years), colors='gray', linestyles='solid')

# Shade Landsat periods
plt.axvspan(ls5_period[0], ls5_period[1], color='lightgray', alpha=0.6)   # light gray
plt.axvspan(ls7_period[0], ls7_period[1], color='darkgray', alpha=0.6)    # darker gray
plt.axvspan(ls8_period[0], ls8_period[1], color='dimgray', alpha=0.6)     # darkest gray

# Plot blue dots
plt.plot(blue_dots, [1] * len(blue_dots), 'ko')

# Landsat labels (below timeline)
plt.text((ls5_period[0] + ls5_period[1]) / 2, 0.96, "Landsat 5", ha='center', fontsize=14, color='black')
plt.text((ls7_period[0] + ls7_period[1]) / 2, 0.96, "Landsat 7", ha='center', fontsize=14, color='black')
plt.text((ls8_period[0] + ls8_period[1]) / 2, 0.96, "Landsat 8", ha='center', fontsize=14, color='black')

# Label blue years
for year in blue_dots:
    plt.text(year, 1.02, str(year), ha='center', fontsize=14, rotation=45)

# X-axis formatting
plt.xticks([1984, 2024], fontsize=14)
plt.xlim(1983, 2025)

# Remove y-axis and frame
plt.yticks([])
plt.ylim(0.93, 1.1)
plt.xlabel("Year", fontsize=14)
plt.title("Landsat Mission Timeline with GE Years: 1984–2024", fontsize=14)

for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.tight_layout()

# Save figure
output_path = "/Users/bz294/Documents/WhiteZone/Figure/Landsat_timeline_corrected.tif"
plt.savefig(output_path, dpi=300, bbox_inches='tight', format='tiff')
plt.show()
```
