```python
######################## Read in a csv and extract all the words in it ############
import csv
import re
```

```python
csv_path = '/Users/bz294/Documents/WhiteZone/Data/Driver_Analysis/Hydro/DBhydro/SW_allsite.csv'  # Replace with your actual file path

with open(csv_path, 'r', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    for i, row in enumerate(reader, start=1):
        print(f"Row {i} has {len(row)} columns: {row}")
```

```python
rows_with_empty_last_two = []

with open(csv_path, 'r', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    for i, row in enumerate(reader, start=1):
        if len(row) >= 2 and row[-1] == '' and row[-2] == '':
            rows_with_empty_last_two.append(i)

print("Rows with last two columns empty:", rows_with_empty_last_two)
```

```python
extracted_data = []

with open(csv_path, 'r', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        if len(row) >= 2 and row[-1] == '' and row[-2] == '':
            # Safely extract columns 0, 3, 6, 9, and 12 if they exist
            selected = []
            for idx in [0, 3, 6, 9, 12]:
                selected.append(row[idx] if len(row) > idx else '')
            extracted_data.append(" | ".join(selected))  # or change separator
extracted_data = extracted_data[1:]

# Flatten the ' | '-joined strings into a single list
flattened_data = []
for row in extracted_data:
    parts = [part.strip() for part in row.split('|') if part.strip() != '']
    flattened_data.extend(parts)
```

```python
all_processed_rows = []

with open(csv_path, 'r', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        if (row[-1] == '' or row[-2] == '') and not (row[-1] == '' and row[-2] == ''):
            combined_row = []
            pre_group = []
            
            for element in row:
                # Check if the element is a single letter, single digit number, or double digit number
                if re.match(r'^[a-zA-Z]$', element):  # Single letter
                    if pre_group:
                        combined_row.append(' '.join(pre_group))
                        pre_group = []
                    continue  # Exclude the letter
                
                if re.match(r'^\d{1}$', element):  # Single digit number
                    if pre_group:
                        combined_row.append(' '.join(pre_group))  # Combine previous elements
                        pre_group = []
                    continue  # Skip the single digit number
                
                if re.match(r'^\d{2}$', element):  # Double digit number
                    if pre_group:
                        combined_row.append(' '.join(pre_group))
                        pre_group = []
                    continue  # Skip the double digit number

                # If the element is part of a group (not a single letter/number), add it to pre_group
                pre_group.append(element)

            # Add any remaining group if it's not empty
            if pre_group:
                combined_row.append(' '.join(pre_group))

            all_processed_rows.append(combined_row)

all_processed_rows = [
[elem for elem in row if elem.strip() != '']
for row in all_processed_rows
]
              
special_flatten = [item for row in all_processed_rows for item in row]

for processed in special_flatten:
    print(processed)
```

```python
concatenated_data = flattened_data + special_flatten
output_csv_path = '/Users/bz294/Documents/WhiteZone/Data/Driver_Analysis/concatenated_output.csv'

# Write the concatenated data into the CSV
with open(output_csv_path, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(concatenated_data)  # Write the entire list as a single row

print(f"Data has been written to {output_csv_path}")
```
